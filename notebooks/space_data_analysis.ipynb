# Voyage Beyond: Space Data Analysis
This notebook loads the curated planetary dataset, checks data quality, computes descriptive statistics, and visualizes planetary diameters. Verify and cite the source in the final report. Blank cells represent unavailable values, not zero.

```python
from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

DATA_PATH = Path('..') / 'data' / 'planets.csv'
planets = pd.read_csv(DATA_PATH)
planets.head()
```

## 1. Inspect the dataset
Check the number of records, column types, and missing values before analyzing measurements.

```python
print('Shape:', planets.shape)
print('Categories:', planets['category'].value_counts().to_dict())
display(planets.isna().sum().sort_values(ascending=False).to_frame('missing_values'))
planets.info()
```

## 2. Descriptive statistics
Pandas excludes missing values from numeric summaries.

```python
numeric_columns = ['diameter_km', 'gravity_m_s2', 'orbital_period_days', 'mean_temperature_c', 'distance_from_sun_million_km']
for column in numeric_columns:
    planets[column] = pd.to_numeric(planets[column], errors='coerce')
display(planets[numeric_columns].describe().T)
```

## 3. Compare diameters
The Moon and Pluto are included in this wider celestial-object dataset. Interpret categories carefully.

```python
diameters = planets.dropna(subset=['diameter_km']).sort_values('diameter_km', ascending=False)
plt.figure(figsize=(10, 5))
plt.bar(diameters['name'], diameters['diameter_km'])
plt.title('Diameter of selected Solar System objects')
plt.xlabel('Object')
plt.ylabel('Diameter (km)')
plt.xticks(rotation=35, ha='right')
plt.tight_layout()
plt.show()
```

## 4. Orbital period and distance
This plot compares two numerical properties. It describes these selected objects and does not establish causation. The Moon's distance is from Earth, so exclude it from this plot.

```python
plot_data = planets[planets['category'] == 'Planet'].dropna(subset=['distance_from_sun_million_km', 'orbital_period_days'])
plt.figure(figsize=(8, 5))
plt.scatter(plot_data['distance_from_sun_million_km'], plot_data['orbital_period_days'])
for _, row in plot_data.iterrows():
    plt.annotate(row['name'], (row['distance_from_sun_million_km'], row['orbital_period_days']), fontsize=8)
plt.title('Orbital period versus distance from the Sun')
plt.xlabel('Average distance (million km)')
plt.ylabel('Orbital period (days)')
plt.tight_layout()
plt.show()
```
